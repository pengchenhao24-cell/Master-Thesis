```python
!pip install gensim

# 导入库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from gensim.utils import simple_preprocess
from gensim import corpora
from gensim.models import LdaModel, CoherenceModel

# ---------------------------
# 1. 读取数据
# ---------------------------

# 如果你是手动上传文件，把下面这两行取消注释：
# from google.colab import files
# uploaded = files.upload()  # 选择 combined_comments_final.csv

# 读取 CSV（文件名要和你上传的一致）
df = pd.read_csv("combined_comments_final.csv")

print("列名：", df.columns)

# 用 clean_text 作为文本列（如果想用 text，可以改成 df['text']）
text_col = "clean_text"

# 去掉缺失，并转成字符串
texts_raw = df[text_col].fillna("").astype(str).tolist()

print("总评论数：", len(texts_raw))
print("示例原始文本：", texts_raw[0][:100])

# ---------------------------
# 2. 文本预处理
# ---------------------------

def preprocess(text):
    # simple_preprocess: 全部小写，去标点，按词切分
    return simple_preprocess(text, deacc=True)

texts = [preprocess(t) for t in texts_raw]

print("示例分词结果：：", texts[0][:15])

# ---------------------------
# 3. 构造词典和语料
# ---------------------------

dictionary = corpora.Dictionary(texts)

# 过滤极端词（根据需要可以调参）
dictionary.filter_extremes(
    no_below=20,   # 至少出现在 20 个文档中
    no_above=0.5   # 出现在超过 50% 文档中的词删除
)

corpus = [dictionary.doc2bow(t) for t in texts]

print("词典大小：", len(dictionary))
print("语料文档数：", len(corpus))

# ---------------------------
# 4. 定义函数：计算不同 K 的 Coherence
# ---------------------------

def compute_coherence_values(dictionary, corpus, texts, start=3, limit=10, step=1):
    """
    计算从 start 到 limit（包含）不同主题数 K 的 LDA coherence（c_v）
    返回：
    - model_list: 每个 K 对应的 LDA 模型
    - coherence_values: 每个 K 对应的 coherence 分数
    """
    model_list = []
    coherence_values = []

    for k in range(start, limit + 1, step):
        print(f"\n正在训练 K = {k} 的 LDA 模型...")
        lda_model = LdaModel(
            corpus=corpus,
            id2word=dictionary,
            num_topics=k,
            random_state=42,
            chunksize=2000,
            passes=10,
            alpha='auto',
            per_word_topics=False
        )
        model_list.append(lda_model)

        coherencemodel = CoherenceModel(
            model=lda_model,
            texts=texts,
            dictionary=dictionary,
            coherence='c_v'
        )
        coherence = coherencemodel.get_coherence()
        coherence_values.append(coherence)
        print(f"K={k} 的 coherence = {coherence:.4f}")

    return model_list, coherence_values

# ---------------------------
# 5. 计算 K=3 到 K=10 的 Coherence
# ---------------------------

start_k = 3
end_k = 9

model_list, coherence_values = compute_coherence_values(
    dictionary=dictionary,
    corpus=corpus,
    texts=texts,
    start=start_k,
    limit=end_k,
    step=1
)

# ---------------------------
# 6. 画出 Coherence 曲线 & 打印结果
# ---------------------------

x = list(range(start_k, end_k + 1))

plt.figure(figsize=(7, 5))
plt.plot(x, coherence_values, marker="o")
plt.xlabel("Number of Topics (K)")
plt.ylabel("Coherence Score (c_v)")
plt.title("Topic Coherence for Different K")
plt.xticks(x)
plt.grid(True)
plt.tight_layout()
plt.show()

print("\n各个 K 的 Coherence 值：")
for k, c in zip(x, coherence_values):
    print(f"K={k}: coherence={c:.4f}")

# 找出最佳 K（coherence 最大）
best_index = int(np.argmax(coherence_values))
best_k = x[best_index]
best_coherence = coherence_values[best_index]

print(f"\n最佳主题数 K = {best_k}, 对应的 Coherence = {best_coherence:.4f}")

# ---------------------------
# 7. 查看最佳模型的主题词
# ---------------------------

best_model = model_list[best_index]

print(f"\n===== 最佳模型（K={best_k}）的主题词 =====")
for t in range(best_k):
    print(f"\n主题 {t}:")
    print(best_model.print_topic(t, topn=10))

# 如需保存模型和词典（可选）
best_model.save(f"lda_k{best_k}_model")
dictionary.save("lda_dictionary.dict")

print("\n全部完成 ✅")
```

```python
# ============================================
# 4.4 用户回复网络 SNA 分析 —— Colab 完整代码
# 节点：authorName
# 边：authorName -> isReplyToName（回复关系）
# ============================================

!pip install networkx matplotlib

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# 1. 读取数据
df = pd.read_csv("combined_comments_final.csv", low_memory=False)

print("数据行数：", len(df))
print("列名：", df.columns.tolist())

# 2. 只保留有回复对象的行（即确实是回复，而不是顶层评论）
replies = df[df["isReplyTo"].notna()].copy()
print("回复评论条数：", len(replies))

# 3. 构建边表：source = authorName, target = isReplyToName
edges_df = replies[["authorName", "isReplyToName"]].dropna().copy()

# 去掉明显无效或自回复
edges_df = edges_df[edges_df["authorName"] != edges_df["isReplyToName"]]

print("有效边数量（行）：", len(edges_df))
print(edges_df.head())

# 4. 构建有向图
G = nx.DiGraph()

for _, row in edges_df.iterrows():
    src = str(row["authorName"])
    tgt = str(row["isReplyToName"])
    if src and tgt and src != "nan" and tgt != "nan":
        if G.has_edge(src, tgt):
            G[src][tgt]["weight"] += 1
        else:
            G.add_edge(src, tgt, weight=1)

print("\n网络节点数：", G.number_of_nodes())
print("网络边数：", G.number_of_edges())

# 5. 计算入度、出度及统计特征
in_degrees = dict(G.in_degree(weight="weight"))
out_degrees = dict(G.out_degree(weight="weight"))

in_vals = list(in_degrees.values())
out_vals = list(out_degrees.values())

total_replies = sum(in_vals)

import numpy as np
import statistics as stats

mean_in = stats.mean(in_vals)
median_in = np.median(in_vals)
mean_out = stats.mean(out_vals)
median_out = np.median(out_vals)
max_in = max(in_vals)
max_out = max(out_vals)

zero_in = sum(1 for v in in_vals if v == 0)
gt5 = sum(1 for v in in_vals if v > 5)
gt20 = sum(1 for v in in_vals if v > 20)

zero_in_pct = zero_in / G.number_of_nodes() * 100
gt5_pct = gt5 / G.number_of_nodes() * 100
gt20_pct = gt20 / G.number_of_nodes() * 100

print("\n===== 入度统计（收到多少次回复） =====")
print("平均入度：", round(mean_in, 3))
print("入度中位数：", median_in)
print("最大入度：", max_in)
print(f"入度 = 0 的节点数：{zero_in} ({zero_in_pct:.1f}%)")
print(f"入度 > 5 的节点数：{gt5} ({gt5_pct:.1f}%)")
print(f"入度 > 20 的节点数：{gt20} ({gt20_pct:.1f}%)")

print("\n===== 出度统计（向多少人回复） =====")
print("平均出度：", round(mean_out, 3))
print("出度中位数：", median_out)
print("最大出度：", max_out)

# 6. 找出入度最高的前 10 个用户
top_in = sorted(in_degrees.items(), key=lambda x: x[1], reverse=True)[:10]

print("\n===== 入度最高的前 10 个账号（被回复次数最多） =====")
for rank, (node, deg) in enumerate(top_in, start=1):
    print(f"{rank}. {node} —— 收到回复 {deg} 次")

# 7. 可视化：以入度最高的节点为中心的 ego 网络（示意）
top_node = top_in[0][0]
print("\n入度最高的节点：", top_node)

ego = nx.ego_graph(G, top_node, radius=1, undirected=True)

plt.figure(figsize=(10, 8))
pos = nx.spring_layout(ego, k=0.5, seed=42)

# 节点大小根据度
node_sizes = [ego.degree(n) * 50 for n in ego.nodes()]

nx.draw_networkx_nodes(ego, pos, node_size=node_sizes, node_color="skyblue")
nx.draw_networkx_edges(ego, pos, alpha=0.6)
nx.draw_networkx_labels(ego, pos, font_size=8)

plt.title(f"Ego Network of Top Replied User: {top_node}")
plt.axis("off")
plt.show()

print("\nSNA 分析完成 ✅")
```

```python
import pandas as pd
import re
from collections import Counter

df = pd.read_csv("combined_comments_final.csv", low_memory=False)

# ===== 你定义的四大主题词典 =====
topic_dict = {
    "kobe_memorial": ["kobe", "mamba", "rip", "legend", "bryant", "miss", "peace"],
    "historical_game": ["curry", "kyrie", "finals", "history", "back", "watching", "video", "warriors"],
    "light_reaction": ["lol", "good", "watch", "people", "bro", "dunk", "man", "like", "time"],
    "goat_debate": ["goat", "lebron", "jordan", "best", "greatest", "james", "ever", "one"]
}

# ===== 找到入度最高的前 5 个核心用户 =====
replies = df[df["isReplyToName"].notna()]
in_degree = replies["isReplyToName"].value_counts()
top_users = list(in_degree.head(5).index)

print("入度最高的前 5 位用户：")
print(top_users)

# ===== 工具函数 =====
def tokenize(text):
    return re.findall(r"[a-zA-Z]+", str(text).lower())

def analyze_user(core_user):
    replies_to_user = df[df["isReplyToName"] == core_user]["clean_text"].astype(str).tolist()

    tokens = []
    for t in replies_to_user:
        tokens.extend(tokenize(t))
    freq = Counter(tokens)

    topic_counts = {}
    for topic, words in topic_dict.items():
        topic_counts[topic] = sum(freq[w] for w in words)

    return topic_counts, len(replies_to_user)

# ===== 对每个核心用户做分析 =====
results = {}

for user in top_users:
    topic_counts, reply_count = analyze_user(user)
    results[user] = {
        "total_replies": reply_count,
        **topic_counts
    }

# ===== 输出结果 =====
print("\n===== 核心用户主题分布统计 =====")
for user, stats in results.items():
    print(f"\n--- {user} ---")
    print(f"总回复数：{stats['total_replies']}")
    print("主题分布：")
    for topic, count in stats.items():
        if topic != "total_replies":
            print(f"  {topic:15} : {count}")
```

```python
df_topic = pd.DataFrame(results).T   # 行是用户，列是各项统计
# 只保留四个话题列
topic_cols = list(topic_dict.keys())
df_topic_plot = df_topic[topic_cols]

print("\n用于绘图的表：")
print(df_topic_plot)

# ============= 7. 画堆叠柱状图 =============
plt.figure(figsize=(10, 6))

bottom = None
colors = ["#4c72b0", "#55a868", "#c44e52", "#8172b3"]  # 颜色可改可不改

for i, topic in enumerate(topic_cols):
    if bottom is None:
        plt.bar(df_topic_plot.index, df_topic_plot[topic], label=topic, color=colors[i])
        bottom = df_topic_plot[topic].copy()
    else:
        plt.bar(df_topic_plot.index, df_topic_plot[topic], bottom=bottom, label=topic, color=colors[i])
        bottom += df_topic_plot[topic]

plt.xticks(rotation=45, ha="right")
plt.ylabel("Word frequency (per topic)")
plt.title("Topic-related Word Frequency in Replies to Top-5 Core Users")
plt.legend(title="Topic category")
plt.tight_layout()
plt.show()
```

```python
# =========================
# 0) 安装依赖（Colab 需要；本地已有可跳过）
# =========================
!pip -q install vaderSentiment

import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# =========================
# 1) 读数据
# =========================
df = pd.read_csv("combined_comments_final.csv", low_memory=False)

# 选一个文本列：优先 clean_text，其次 text
text_col = "clean_text" if "clean_text" in df.columns else "text"
if text_col not in df.columns:
    raise ValueError("找不到文本列：需要 clean_text 或 text")

# =========================
# 2) 解析时间（你这份数据是 'YYYY/M/D H:MM' 这种格式，to_datetime 能识别）
#    这一步是你之前“为什么只到2023”的最常见原因：时间解析失败或被错误过滤
# =========================
df["publishedAt_dt"] = pd.to_datetime(df["publishedAt"], errors="coerce")
df = df.dropna(subset=["publishedAt_dt"]).copy()

# =========================
# 3) 只保留“科比相关”，但扩大范围
#    说明：
#    - 用“词边界 + 常见短语”避免误伤
#    - 不建议用纯数字(24/8)当关键词，会引入大量无关评论
# =========================
KOBE_REGEX = re.compile(
    r"""
    (?:\bKobe\b)|
    (?:\bBryant\b)|
    (?:\bKobe\s+Bryant\b)|
    (?:\bMamba\b)|
    (?:\bBlack\s+Mamba\b)|
    (?:\bMamba\s+Mentality\b)|
    (?:\bRIP\b.*\bKobe\b)|
    (?:\bRest\s+in\s+peace\b.*\bKobe\b)|
    (?:\bGigi\b)|
    (?:\bGianna\b)|
    (?:\bKB24\b)|
    (?:\bBean\b.*\bBryant\b)
    """,
    re.IGNORECASE | re.VERBOSE
)

df[text_col] = df[text_col].fillna("").astype(str)
kobe_df = df[df[text_col].str.contains(KOBE_REGEX, na=False)].copy()

print("总评论数:", len(df))
print("科比相关评论数(扩大范围后):", len(kobe_df))

# =========================
# 4) （可选）情绪分析：不影响你要的“按月对比图”
#    如果你只要“评论数量按月趋势”，可以不看这列
# =========================
analyzer = SentimentIntensityAnalyzer()
kobe_df["sent_compound"] = kobe_df[text_col].apply(lambda t: analyzer.polarity_scores(t)["compound"])

# =========================
# 5) 按“年-月”聚合：你要的是看每年1月是否抬升，所以用 month=1..12 做横轴
# =========================
kobe_df["year"] = kobe_df["publishedAt_dt"].dt.year
kobe_df["month"] = kobe_df["publishedAt_dt"].dt.month

monthly_counts = (
    kobe_df.groupby(["year", "month"])
           .size()
           .reset_index(name="count")
)

# 做一个 12(月) x N(年) 的透视表；缺的月份补0
pivot_counts = (
    monthly_counts.pivot(index="month", columns="year", values="count")
                  .reindex(range(1, 13))
                  .fillna(0)
                  .sort_index()
)

# =========================
# 6) 输出：按年对比“每月评论数量”折线图（你要的图）
#    每条线 = 一个年份；横轴 = 1~12月；纵轴 = 科比相关评论数
# =========================
plt.figure(figsize=(11, 6))

for year in pivot_counts.columns:
    plt.plot(pivot_counts.index, pivot_counts[year].values, marker="o", linewidth=1)

plt.xticks(range(1, 13))
plt.xlabel("Month (1=Jan ... 12=Dec)")
plt.ylabel("Kobe-related comment count")
plt.title("Monthly trend of Kobe-related YouTube comments (lines = years)")
plt.grid(True, alpha=0.3)
plt.legend([str(y) for y in pivot_counts.columns], title="Year", ncol=2, fontsize=9)
plt.tight_layout()
plt.show()

# =========================
# 7) （可选）快速检查：每年1月的评论数，方便你直接验证“每年1月是否增加”
# =========================
jan_series = pivot_counts.loc[1].sort_index()
print("\n每年 1 月（Jan）的科比相关评论数：")
print(jan_series.to_string())
```
